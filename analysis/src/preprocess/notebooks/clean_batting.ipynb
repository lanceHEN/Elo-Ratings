# Cleaning batting
This notebook provides a clean version of the raw batting.csv file.
If does the following:
1. Add a 'season' column
2. Save to a new csv, `.data/clean/batting_cleaned.csv`

```python
pd.set_option('display.max_columns', None)
```

```python
batting = pd.read_csv(Path.cwd() / "analysis" / "data" / "raw" / "batting.csv")
batting.head()
```

## 1. Add 'season' column

```python
batting['season'] = batting['date'].apply(lambda x: int(str(x)[:4]))
batting.head()
```

## 2. Save to `.csv`

```python
batting.to_csv(Path.cwd() / "analysis" / "data" / "clean" / "batting_clean.csv", index=False)
```
