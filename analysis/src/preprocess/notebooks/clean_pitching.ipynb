# Cleaning pitching
This notebook provides a clean version of the raw pitching.csv file.
If does the following:
1. Add a 'season' column
2. Save to a new csv, `.data/clean/pitching_cleaned.csv`

```python
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv(Path.cwd() / "analysis" / "data" / "raw" / "pitching.csv")
df.head()
```

## 1. Add 'season' column

```python
df['season'] = df['date'].apply(lambda x: int(str(x)[:4]))
df.head()
```

## 2. Save to `.csv`

```python
df.to_csv(Path.cwd() / "analysis" / "data" / "clean" / "pitching_clean.csv", index=False)
```
