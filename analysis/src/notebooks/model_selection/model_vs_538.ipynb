# Model vs 538

## Evaluate my Model

```python
bce, accuracy = evaluate_elo_prob_func(games, predict_lr, K=3, margin_of_victory_column='adjustedmarginofvictory', skip_first_n=0)
print(f"BCE: {bce}")
print(f"Accuracy: {accuracy}")
```

```python
bce, accuracy = evaluate_elo_prob_func(games, predict_lr, K=3, margin_of_victory_column='adjustedmarginofvictory', skip_first_n=0, years=set(range(2021,2026)))
print(f"BCE: {bce}")
print(f"Accuracy: {accuracy}")
```
