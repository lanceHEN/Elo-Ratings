# Cleaning team_stats
This notebook provides a clean version of the raw teamstats.csv file.
If does the following:
1. Add a 'season' column
2. Filter for 'value' rows
3. Save to a new csv, `.data/{schedule_name}_cleaned.csv`

```python
pd.set_option('display.max_columns', None)
```

```python
teamstats = pd.read_csv(f'teamstats.csv')
teamstats.head()
```

## 1. Add 'season' column

```python
teamstats['season'] = teamstats['date'].apply(lambda x: int(str(x)[:4]))
teamstats.head()
```

## 2. Filter for 'value' rows

```python
teamstats = teamstats[teamstats['stattype'] == 'value']
```

## 2. Save to `.csv`

```python
teamstats.to_csv(f'./teamstats_clean.csv', index=False)
```
